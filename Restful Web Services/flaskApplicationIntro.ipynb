```python
from flask import Flask, request, jsonify
app = Flask(__name__)
```

```python
students = [
    {"id": 1, "name": "Jon", "grade": 88, "letterGrade": "A"},
    {"id": 2, "name": "Ann", "grade": 77, "letterGrade": "B"},
    {"id": 3, "name": "Tod", "grade": 66, "letterGrade": "C"}
]
```

```python
#Creating an id for the student at the next index.
def _find_next_id():
    return max(stu["id"] for stu in students) + 1
```

```python
@app.get("/students")
def get_students():
    return jsonify(students)
```

```python
@app.post("/students")
def add_students():
    if request.is_json:
        stu = request.get_json()
        stu["id"] = _find_next_id()
        students.append(stu)
        return stu
    return {"error":"Request must be JSON"}
```

```python
if __name__ == '__main__':
    app.run()
```
