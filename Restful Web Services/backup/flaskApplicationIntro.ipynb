```python
#This is the start
from flask import Flask
app = Flask(__name__)   #Put 2 __ on each side. This is the start of your Flask application. 
```

```python
@app.get('/')
def Hello_world():
    return "Hello Hello Hello!"
```

```python
#This is the end
if  __name__ == '__main__':
    app.run()
#This section should continue to run until you stop your application.
```
