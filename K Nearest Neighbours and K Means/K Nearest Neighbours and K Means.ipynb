This notebook demonstrates the implementation of the K-Means clustering algorithm using the Seaborn Palmer Penguins dataset to segment observations into species clusters.

## Workflow Overview
- Data Loading & Preprocessing
- Categorical Encoding & Missing Value Handling
- K-Means Model Fitting ($k=3$)
- Evaluation & Cluster Analysis


The goal is to apply K-Means clustering to partition the Palmer Penguins dataset into three distinct groups corresponding to the three target species.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

penguins = sns.load_dataset('penguins')
```

```python
#Possible species
penguins['species'].unique()
```

We are going to have to replace the species as well as K Means won't be able to process the object type.

```python
penguins.replace('Adelie', 0, inplace=True)
penguins.replace('Chinstrap', 1, inplace=True)
penguins.replace('Gentoo', 2, inplace=True)

penguins.replace('Torgersen', 0, inplace=True)
penguins.replace('Biscoe', 1, inplace=True)
penguins.replace('Dream', 2, inplace=True)

penguins.replace('Male', 0, inplace=True)
penguins.replace('Female', 1, inplace=True)

penguins.dropna(axis=0, inplace=True)
```

```python
from sklearn.cluster import KMeans
```

We are going to divide our clusters into 3 groups (Each species)

```python
kmeans = KMeans(n_clusters=3)
```

```python
kmeans.fit(penguins.drop('species',axis=1))
```

After trying to process the different iterations, the cluster centers are calculated below.

```python
kmeans.cluster_centers_
```

```python
from sklearn.metrics import classification_report
print(classification_report(penguins['species'],kmeans.labels_))
```
